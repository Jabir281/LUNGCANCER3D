# Download Trained Model
This notebook helps you verify and download your trained model (`outputs/checkpoints/best_model.pth`) from the Jupyter environment.

```python
from pathlib import Path
from IPython.display import FileLink, display

# If running from the notebooks/ folder, hop to project root
project_root = Path.cwd()
if project_root.name == "notebooks":
    project_root = project_root.parent
```

```python
# Define the model path
model_path = project_root / "outputs" / "checkpoints" / "best_model.pth"

# Check if file exists and print size
if model_path.exists():
    size_mb = model_path.stat().st_size / (1024 * 1024)
    print(f"Found model file: {model_path}")
    print(f"Size: {size_mb:.2f} MB")
else:
    print(f"Error: File {model_path} not found. Did training finish?")
```

```python
# Generate download link
if model_path.exists():
    display(FileLink(str(model_path)))
```
